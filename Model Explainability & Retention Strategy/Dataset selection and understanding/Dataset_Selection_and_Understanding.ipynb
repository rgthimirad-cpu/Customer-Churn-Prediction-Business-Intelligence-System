## Identifier Variable

`customerID` is the identifier variable.

It uniquely identifies individual customers and is not a meaningful predictive feature for customer churn. Therefore, it should not be used as an input feature during model training.

The identifier will be retained in the raw dataset for record identification but excluded from the machine learning feature set in later stages.

```python
# Define the identifier and target variable
identifier_columns = []

target_column = "Churn Label"

# Select all columns except the target variable as features
feature_columns = [
    column for column in df.columns
    if column not in identifier_columns + [target_column]
]

# Display the identifier, target, and feature columns
print("Identifier:")
print(identifier_columns)

print("\nTarget:")
print(target_column)

print("\nFeatures:")
for feature in feature_columns:
    print(feature)

# Display the total number of features
print("\nNumber of features:", len(feature_columns))
```

```python
# Identify numerical and categorical features from the selected feature columns

numerical_features = df[feature_columns].select_dtypes(
    include="number"
).columns.tolist()

categorical_features = df[feature_columns].select_dtypes(
    include="object"
).columns.tolist()

# Display the numerical and categorical features
print("Numerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)
```

## Business Context

Customer churn refers to customers discontinuing their services with a company.

In the telecommunications industry, customer churn can result in:

- Loss of recurring revenue
- Increased customer acquisition costs
- Reduced customer lifetime value
- Loss of long-term customers
- Increased pressure on customer retention strategies

The purpose of the dataset is to identify customer characteristics and service/account patterns associated with churn.

The machine learning problem can therefore be formulated as a binary classification problem:

- `Churn Label = Yes` → Customer churned
- `Churn Label = No` → Customer remained with the company

The future machine learning model will use customer demographic, service, contract, and billing information to predict the probability that a customer will churn.

```python
# Assign a role to each column: Identifier, Target, Excluded, or Feature
data_types = []

for column in df.columns:
    if column == "Churn Label":
        variable_role = "Target"
    elif column in ["Churn Value", "Churn Score", "Churn Reason"]:
        variable_role = "Excluded"
    else:
        variable_role = "Feature"

    data_types.append(variable_role)

# Add the variable role and data type to the data dictionary
data_dictionary_df["Role"] = data_types

data_dictionary_df["Data Type"] = [
    df[column].dtype for column in df.columns
]

# Display the completed data dictionary
data_dictionary_df
```

Conclusion

The IBM Telco Customer Churn dataset was selected as the final dataset for the Customer Churn Prediction and Business Intelligence System.

The dataset contains 7,043 customer records and 28 columns, with each record representing an individual telecommunications customer. It consists of:

1 record variable: Count
Predictive features covering customer demographics, account information, subscribed telecommunications services, contract details, payment methods, and billing information
1 target variable: Churn Label

The predictive features represent customer demographics, account information, subscribed telecommunications services, contract details, payment methods, and billing information.

The target variable, Churn Label, is a binary categorical variable with two possible outcomes:

Yes — Customer churned
No — Customer did not churn

Therefore, the dataset is suitable for a supervised binary classification problem.

The raw dataset has been preserved without preprocessing or transformation. Initial data-quality checks identified potential issues, particularly blank or invalid values in the Total Charges column, which will be addressed during the preprocessing stage.

The raw dataset and data dictionary have been prepared for the next stages of the project, including data cleaning, preprocessing, exploratory data analysis (EDA), feature engineering, and machine learning model development.
